```python

from cellpose import plot
import tifffile

#model = models.CellposeModel(pretrained_model='C:/Users/Kevin Tsia/.cellpose/models/cyto')
model=models.CellposeModel(model_type='cyto2')

for file in tqdm(files):
    t1 = time.time()
    print(file)
    img = imread(file)
    
    masks, flows, styles = model.eval(img, diameter=diam, do_3D=False, resample=True, progress=True)    
    #Save the label
    label_name = file.replace(".tif", "_label.tif")
    cv2.imwrite(label_name, masks.astype(np.uint8))
    t2 = time.time()
    time_elapsed = (t2 - t1) / 60
    print('Time spent in current image: %s minutes' % round(time_elapsed,1))
    print('------')
    
```
